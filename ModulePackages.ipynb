# 1. Creating Custom Modules in Python

## Definition

A **module** is a Python file (`.py`) that contains <u>functions</u>, <u>variables</u>, <u>classes</u>, and <u>executable code</u> that can be reused in other Python programs.

---


## Key Points

- A module helps organize code into separate files.
- It promotes **code reusability** and reduces duplication.
- Modules make large programs easier to maintain.
- Python provides both **built-in modules** (e.g., `math`, `random`) and **user-defined modules**.
- A module can contain variables, functions, classes, and executable statements.
- Modules are imported using the `import` statement.

---


## Example 1: Creating a Custom Module

### File: `calculator.py`



````python
# calculator.py

def add(a, b):
    return a + b

def multiply(a, b):
    return a * b
````



### File: `main.py`



````python
# main.py

import calculator

print(calculator.add(5, 3))
print(calculator.multiply(4, 2))
````

## Example 2: Import Specific Functions



````python
from calculator import add

print(add(10, 20))
````

# 2. Creating Packages in Python

## Definition

A **package** is a collection of related Python modules organized in a directory hierarchy. Packages help structure large applications into manageable components.

---


## Key Points

- A package is simply a folder containing Python modules.
- Packages improve project organization and readability.
- A package may contain sub-packages.
- The `__init__.py` file is a special Python file placed inside a folder to define and initialize a Python package. It is commonly used, although it is optional in modern Python. When the package is imported for the first time, the code inside `__init__.py` is executed.
- Packages allow hierarchical naming such as `mypackage.module1`.
- Large projects commonly use packages to separate functionality.

---


## Package Structure



```text
myproject/
│
├── main.py
│
└── mypackage/
    ├── __init__.py
    ├── math_utils.py
    └── string_utils.py
```


## Example 1: Module Inside a Package


### File: `math_utils.py`

````python
def square(x):
    return x * x
````



### File: `main.py`



````python
from mypackage import math_utils

print(math_utils.square(5))
````

````python
from mypackage import math_utils as m

print(m.square(5))
````

## Example 2: Import Specific Function



````python
from mypackage.math_utils import square as s

print(s(6))
````

## Example 3: Package with Sub-Package

### Directory Structure

```text
company/
│

│
└── hr/

    └── employee.py
```




### File: `employee.py`

# 3. Introduction to `pip`

## Definition

**pip** (Pip Installs Packages) is Python's package manager used to <u>install</u>, <u>upgrade</u>, <u>remove</u>, and <u>manage third-party libraries</u> and packages.

---


## Key Points

- `pip` allows access to thousands of packages from the Python Package Index (PyPI is online repository where Python packages are published and shared).
- It simplifies software installation and dependency management.
- Most Python distributions (Interpreter, Standard libraries, Tools such as `pip`,  Additional tools or packages) include `pip` by default.
- Packages installed via `pip` can be imported directly into Python programs.
- `pip` can install, update, uninstall, and list packages.
- It helps developers reuse existing libraries instead of writing everything from scratch.

---

# Quick Comparison

| Feature | Module | Package | `pip` |
|----------|---------|----------|-------|
| **What is it?** | Single Python file | Collection of modules | Package manager |
| **Extension** | `.py` file | Folder containing modules | Command-line tool |
| **Purpose** | Reuse code | Organize large projects | Install/manage libraries |
| **Example** | `calculator.py` | `mypackage/` | `pip install numpy` |
| **Contains** | Functions, Classes, Variables | Multiple modules and sub-packages | External packages |

---


# Summary

- **Module** -> A single Python file containing reusable code.
- **Package** -> A directory that groups related modules and sub-packages.
- **pip** -> Python's package management tool used to install and manage external libraries.

---
