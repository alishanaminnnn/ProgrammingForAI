# Quadratic Equation
For the quadratic equation $ax2 + bx + c = 0$, the discriminant is: \
$D = b^2 − 4ac$\
Write a function discriminant(a, b, c) that returns D.\
Write another function quadratic_roots(a, b, c) that uses discriminant(...) and
returns:\
two roots if D > 0, using $x = (−b ± D) / (2a)$
one repeated root if $D = 0$
the string "No real roots" if & $D < 0$ Include a short docstring in each function. Test with $ a = 1, b = −5, c = 6 $ and print the result.

```python
def discriminant(a,b,c):
    """Calculates and returns the discriminant (D) of a quadratic equation."""
    D=(b**2)-(4*a*c)
    return D
def quadratic_roots(a,b,c):
    """
    Calculates the roots of a quadratic equation using the discriminant.
    Returns two roots, one root, or a message indicating no real roots.
    """
    
    D=discriminant(a,b,c)
    if D>0:
        negative=(-b-(D**0.5))/(2*a)
        positive=(-b+(D**0.5))/(2*a)
        return positive,negative
    elif D == 0:
        root = -b / (2 * a)
        return root
    else:
        return "No real Roots"

a=1
b=-5
c=6
result=quadratic_roots(a,b,c)
print(result)
```
