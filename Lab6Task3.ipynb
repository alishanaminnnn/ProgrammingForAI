```python
marks = {"Ali": 85, "Sara": 92, "Hamza": 67, "Zara": 74, "Omar": 49,
"Hina": 91, "Usman": 45}
```

```python
def grade(m):
    if m >= 85:
        return "A"
    elif m >= 70:
        return "B"
    elif m >= 50:
        return "C"
    else:
        return "F"
```

```python
print("----RESULT----")
count=0
for i,j in marks.items():
    print("Name: ",i)
    print("Marks: ",j)
    print("grade: ",grade(j))
    if grade(j)!='F':
        count=count+1
    print()
    
```

```python
print("Pass Percentage: ",(count/len(marks))*100)
```
