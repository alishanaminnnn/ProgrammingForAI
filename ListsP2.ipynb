## Problem 2
A rectangular matrix (list of equal-length rows) is given. Build and print its transpose (rows become columns). Do not modify the original.

```python
matrix=[[1,2,3],
        [4,5,6],
        [7,8,9]]
temppp=[[1,2,3],
        [4,5,6],
        [7,8,9]]
rows=len(matrix)
cols=len(matrix[0])
for i in range(rows):
    for j in range (cols):
        matrix[i][j]=temppp[j][i]
        
print(matrix)       
```
