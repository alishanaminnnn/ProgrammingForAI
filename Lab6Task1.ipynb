```python
readings = [("Mon", 31.5), ("Tue", 34.0), ("Wed", 29.5), ("Thu", 36.5),
("Fri", 36.5), ("Sat", 33.0), ("Sun", 30.0)]
```

```python
total=0
for days,temp in readings:
    total=total+temp
days=len(readings)
average_temp=total/days
print("Average Tempreture: ",average_temp)
```

```python
maximum=readings[0][1]
for days,temp in readings:
    if temp>maximum:
        maximum=temp
print("1.The maximum tempreture is: ",maximum)
print("2.Days with the hottest temperature:")
for days,temp in readings:
    if temp==maximum:
        print(days)
```

```python
above_average=[]
for days,temp in readings:
    if temp>average_temp:
        above_average.append(days)
print("Days with above average tempreture: ")
print(above_average)
```

```python
tempretures = []
for days, temp in readings:
    tempretures.append(temp)
tempretures.sort(reverse=True)

print(tempretures)
```
