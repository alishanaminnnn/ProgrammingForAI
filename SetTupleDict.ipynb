# Tuples in Python

## Definition

A **tuple** is an ordered, immutable (unchangeable) collection of elements in Python. Tuples can store items of different data types and are enclosed in parentheses `()`.


## Key Points

- **Ordered:** Elements maintain their insertion order.
- **Immutable:** Once created, elements cannot be modified, added, or removed.
- **Allows Duplicates:** Multiple occurrences of the same value are allowed.
- **Can Store Mixed Data Types:** Integers, strings, floats, lists, etc.
- **Faster than Lists:** Due to immutability, tuples are generally more memory-efficient.
- **Useful for Fixed Data:** Ideal when data should not change during program execution.


## Examples

### Creating a Tuple


````python
student = ("Ali", 21, "BSCS")
print(student)
````

### Accessing Elements


````python
colors = ("red", "green", "blue")

print(colors[0])
print(colors[-1])
````

### Tuple Packing and Unpacking


````python
point = (10, 20, 11)

x, y , z= point

print(x)
print(y)
````

### Immutable Nature


````python
numbers = (1, 2, 3)

# numbers[0] = 10    # Error
````

# Sets in Python

## Definition

A **set** is an unordered collection of unique elements. Sets are enclosed in curly braces `{}` and automatically remove duplicate values.


## Key Points

- **Unordered:** Elements do not have a fixed position.
- **Unique Elements:** Duplicate values are automatically removed.
- **Mutable:** Elements can be added or removed.
- **Efficient Membership Testing:** Searching is generally faster than lists.
- **Supports Mathematical Set Operations:** Union, intersection, difference, etc.
- **Cannot Contain Mutable Objects:** Lists and dictionaries cannot be set elements.



## Examples

### Creating a Set


````python
numbers = {1, 2, 3, 4}
print(numbers)
````

### Duplicate Removal


````python
data = {1, 2, 2, 3, 3, 4}

print(data)
````

### Adding and Removing Elements


````python
fruits = {"apple", "banana"}

fruits.add("orange")
fruits.remove("banana")

print(fruits)
````

### Set Operations


````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A | B)   # Union
print(A & B)   # Intersection
print(A - B)   # Difference
````

# Dictionaries in Python

## Definition

A **dictionary** is a collection of key-value pairs. Each key is unique and is used to access its corresponding value. Dictionaries are enclosed in curly braces `{}`.



## Key Points

- **Stores Data as Key-Value Pairs:** Similar to a real-world dictionary.
- **Keys Must Be Unique:** Duplicate keys overwrite previous values.
- **Mutable:** Items can be added, modified, or removed.
- **Fast Lookup:** Values can be accessed quickly using keys.
- **Can Store Mixed Data Types:** Both keys and values can be of different types.
- **Widely Used for Structured Data:** Such as student records, employee information, configuration settings, etc.



## Examples

### Creating a Dictionary

````python
student = {
    "name": "Ali",
    "age": 21,
    "program": "BSCS"
}

print(student)
````

### Accessing Values


````python
student = {
    "name": "Ali",
    "age": 21
}

print(student["name"])
print(student["age"])
````


### Adding and Updating Data


````python
student = {"name": "Ali"}

student["age"] = 21
student["name"] = "Ahmed"

print(student)
````


### Iterating Through a Dictionary


````python
student = {
    "name": "Ali",
    "age": 21
}

for key, value in student.items():
    print(key, ":", value)
````

### Removing an Item


````python
student = {
    "name": "Ali",
    "age": 21
}

del student["age"]

print(student)
````

### Quick Comparison

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| **Syntax** | `[]` | `()` | `{}` | `{key:value}` |
| **Ordered** | Yes | Yes | No | Yes (Python 3.7+) |
| **Mutable** | Yes | No | Yes | Yes |
| **Duplicates** | Yes | Yes | No | Keys: No, Values: Yes |
| **Indexed Access** | Yes | Yes | No | By Key |
| **Main Purpose** | Collection of items | Fixed data | Unique elements | Key-value mapping |


### Examples

```python
my_list = [1, 2, 3]

my_tuple = (1, 2, 3)

my_set = {1, 2, 3}

my_dict = {"name": "Ali", "age": 20}
```

### Dictionaries vs Maps (C++) - for information only

- **Structure:** Both store data as **key–value pairs**, but Python `dict`(keyword) is **hash-table** based, while C++ `map` is typically implemented using a **balanced tree**. (To be discussed in Data Structures)
- **Ordering:** Python dictionaries preserve **insertion order**, while C++ `map` stores elements in **sorted order of keys**.
- **Access Speed:** Dictionary lookup is typically **O(1)** on average, while C++ `map` lookup is **O(log n)**. (To be discussed in Design and Analysis of Algorithms)
- **Syntax:** Python uses `{key: value}`, while C++ uses `map<Key, Value>`.


## Summary

- **Tuple:** Ordered and immutable collection.
- **Set:** Unordered collection of unique elements.
- **Dictionary:** Mutable collection of key-value pairs used for fast data lookup.
