# Euclidean 
The Euclidean distance between two points is:
$d = √((x2 − x1)2 + (y2 − y1)2)$
Write a function distance$(x1, y1, x2=0, y2=0)$ that returns d.
Default $(x2, y2) = (0, 0)$, so one call can measure distance from the origin.
You may use ** 0.5 for the square root.
Call the function for:
1) point (3, 4) from the origin
2) points (1, 2) and (4, 6)
Print both distances.

```python
def distance(x1,y1,x2=0,y2=0):
    d=(((x2-x1))**2 + ((y2-y1)**2))**0.5
    return d
print("Distance 1: ",distance(3,4))
print("Distance 2: ",distance(1,2,4,6))
```
