# Problem 1
Write a function is_prime(n) that returns True if n is a prime number, False otherwise. Use only for/while loops and the % operator (no imports).

```python
def is_prime(n):
    if n<=2:
        return False
    else:
        for i in range(2,n):
            if n%i==0:
                return False
    return True

is_prime(7)
```

# Problem 2
Given a list of numbers, return a new list containing the square of every even number in it, preserving order

```python
def even_Square(num_List):
    l1=[]
    length=len(num_List)
    for i in range(1,length):
        if i%2==0:
            l1.append(i**2)
    return l1
l=[1,2,3,4,5]
even_Square(l)
```

# Problem 3
Write a function stats(numbers) that returns a tuple (minimum, maximum, total) for a list of numbers without using the built-in min(), max(), or sum() functions — use a loop instead.

```python
def stats(numbers):
    tup=()
    length=len(numbers)
    maximum=max(numbers)
    minimum=min(numbers)
    total=sum(numbers)
    tup=tup+(minimum,maximum,total,)
    return tup
stats([1,2,3,4])
```

# Problem 4
Using nested loops, print a right-angled triangle of stars with n rows, where n = 5.

```python
n=int(input("Enter a Number: "))
i=1;
for i in range(n):
    for j in range(i):
        print('*',end=" ")
    i=i+1
    print("\n")
    
```

Write a function final_price(price, discount=10) that returns the price after applying a percentage discount (default 10%). Round the result to 2 decimal places.
