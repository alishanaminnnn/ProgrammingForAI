## Problem 1
A list nums of integers is given. Build two new lists, one with the even numbers and one with the
odd numbers, each keeping the original order. Print the evens list, then the odds list

```python
l1=[4, 7, 10, 3, 8]
l2=[]
l3=[]
for i in l1:
    if i%2==0:
        l2.append(i)
    else:
        l3.append(i)
print(l2) 
print(l3)
```
